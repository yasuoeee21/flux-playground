```python
from llm.api import ContentBuilder, Agent
from PIL import Image
import os
from tqdm import tqdm
```

```python
api_key = '3PxBu2J1g8rrLMvfo3QGsjOwNDXVmXZC'
model = "pixtral-large-2411"
agent = Agent(token=api_key, model=model)
```

```python
img_dir = '/amax/hchuz/architectural_heritage/相似的几个2'
names = [name for name in os.listdir(img_dir) if name.endswith('.png')]
```

```python
for name in tqdm(names):
    image = Image.open(os.path.join(img_dir, name)).convert('RGB')
    contentbuilder = ContentBuilder()
    contentbuilder.add_image(image)
    contentbuilder.add_text('describe the image')
    text = agent.chat(contentbuilder.finish())
    agent.clear()
    with open(os.path.join(img_dir, name.replace('.png', '.txt')), 'w') as f:
        f.write(text)
```
